# Pipeline Silver — Unión Dataset Master

## Descripción
Este notebook consolida los datasets **HIST** y **DIA** en una única tabla Silver final.

## Proceso
1. Lectura de `dataset_master_hist` y `dataset_master_dia`.
2. Unión **FULL JOIN por `DateKey`**.
3. Escritura de la tabla consolidada `dataset_master`.

```python
%run ./Parametros
```

```python
# Tablas origen
table_hist = f"{silver_env}.marcobre_fileserver.dataset_master_hist"
table_dia  = f"{silver_env}.marcobre_fileserver.dataset_master_dia"

# Tabla destino
table_final = f"{silver_env}.marcobre_fileserver.dataset_master"
```

```python
# Leer tablas Silver intermedias
df_hist = spark.table(table_hist)
df_dia  = spark.table(table_dia)
```

```python
# FULL JOIN por DateKey
df_final = (
    df_hist.alias("h")
    .join(
        df_dia.alias("d"),
        on="DateKey",
        how="full"
    )
)
```

```python
# Tabla final
df_final.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable(table_final)
```
