```python
storage_name = "mcstoragelab"

def mount_blob(container):
  if not any(mount.mountPoint == f"/mnt/{container}" for mount in dbutils.fs.mounts()):
    dbutils.fs.mount(
      source = f"wasbs://{container}@{storage_name}.blob.core.windows.net",
      mount_point = f"/mnt/{container}",
      extra_configs = {f"fs.azure.account.key.{storage_name}.blob.core.windows.net":""})
```

```python
#mount_blob("raw")
mount_blob("landing")
mount_blob("prelanding")
```
